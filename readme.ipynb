# 布谷鸟哈希电话号码管理系统 (Cuckoo Hash for Phone Number Management)

## 📌 项目简介

本项目是本科毕业论文《密码学中布谷鸟哈希算法的实现与分析》的配套代码实现。  
设计并实现了一个基于 **SHA-256** 和 **布谷鸟哈希算法** 的电话号码管理系统，通过引入**辅助链表**解决传统布谷鸟哈希插入过程中的死循环问题。

## 🎯 主要工作

- 基于 SHA-256 设计双哈希函数，将11位电话号码映射到两个哈希桶
- 实现布谷鸟哈希的插入（带踢出机制）、查询、删除操作
- 增加回收链表，当踢出次数超过阈值时，将键值对存入链表，**避免哈希表扩容带来的性能损耗**
- 通过平行对照实验（α = 1.0 ~ 5.0），验证了空间利用率与插入/查询效率的权衡

## 📊 实验结果（摘自论文）

| α (空间倍数) | 负载因子 | 死循环率 | 平均插入访问次数 |
|-------------|---------|---------|----------------|
| 1.0         | 0.8385  | 16.15%  | 86.10          |
| 1.8         | 0.5548  | 0.144%  | 2.70           |
| 2.0         | 0.5000  | 0.000%  | 1.18           |

> 当 α ≥ 1.8 时，插入、查询、删除均达到**常数级复杂度**，空间利用率优秀。

## 🛠️ 技术栈

- Python 3.8+
- hashlib（SHA-256）
- 自定义哈希表 + 单链表
